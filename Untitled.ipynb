```python
import json
import pandas as pd
```

```python
datos = []

with open("Enroll Log Sept 29 2025.log", "r") as f:
    for linea in f:
        linea = linea.strip()
        if linea:
            datos.append(json.loads(linea))
```

```python
df = pd.json_normalize(datos)
df.head()
```

```python
columnas = [
    "decode", "detect", "refine", "assess",
    "vector", "search", "verify", "commit", "entire"
]

df_stats = df[columnas]
df_stats.head()
```

```python
df_stats.to_csv("enroll_log_sept_29_2025.csv", index=False)
```

```python
media = df_stats.mean()
media
```

```python
desviacion = df_stats.std()
desviacion
```

```python
moda = df_stats.mode().iloc[0]
moda
```

```python
kurtosis = df_stats.kurtosis()
kurtosis
```

```python
sesgo = df_stats.skew()
sesgo
```

```python
plt.figure(figsize=(10,5))
media.plot(kind="bar")
plt.title("Media de tiempos por etapa")
plt.ylabel("Tiempo (ms)")
plt.xlabel("Proceso")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df_stats.corr(), annot=True, cmap="coolwarm")
plt.title("HeatMap de correlación")
plt.show()
```
